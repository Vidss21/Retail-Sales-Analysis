### Business Insight 1

The West region generated the highest sales ($725K), while the South region recorded the lowest sales. This suggests strong customer demand in the West and potential growth opportunities in the South region.

```python
profit_category = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)

print(profit_category)
```

### Business Insight 2

Technology was the most profitable category, generating approximately $145K in profit. Furniture produced significantly lower profits despite contributing to sales, suggesting lower margins or higher discounting within that category.

```python
monthly_sales = df.groupby('Month')['Sales'].sum().sort_values(ascending=False)

print(monthly_sales)
```

### Business Insight 3

Sales peaked during November and December, indicating strong year-end demand and seasonal purchasing behavior. February recorded the lowest sales, suggesting reduced customer activity during the beginning of the year.

```python
monthly_sales = df.groupby('Month')['Sales'].sum()

plt.figure(figsize=(10,5))
monthly_sales.plot(marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.grid(True)

plt.tight_layout()

plt.savefig('visualizations/monthly_sales_trend.png')

plt.show()
```

```python
correlation = df[['Sales','Profit','Discount']].corr()

print(correlation)
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Discount'], df['Profit'])

plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')

plt.show()
```

### Business Insight 4

Discounts showed a negative correlation with profit (-0.219). Transactions with higher discounts frequently resulted in losses, indicating that aggressive discounting strategies may reduce overall profitability.

## Key Findings

1. The West region generated the highest sales ($725K), while the South region recorded the lowest sales.

2. Technology was the most profitable category, contributing approximately $145K in profit.

3. Sales peaked during November and December, indicating strong seasonal demand.

4. Higher discounts were associated with lower profits, suggesting that excessive discounting negatively impacts profitability.

5. Sales and profit exhibited a moderate positive relationship, indicating that revenue growth generally contributes to increased profitability.

# Conclusion

This project analyzed 9,994 retail transactions to identify sales trends, profitability drivers, and business opportunities. The analysis revealed that the West region generated the highest sales, Technology was the most profitable category, and sales peaked during the year-end months of November and December. Correlation analysis showed that higher discounts were associated with lower profits, highlighting the impact of discounting strategies on profitability. These insights can support data-driven business decisions related to sales planning, product strategy, and revenue optimization.
